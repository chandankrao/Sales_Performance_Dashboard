```python
for column in ['Sales', 'Profit']:
    
    mean = df[column].mean()
    median = df[column].median()
    
    print(f"\n{column}")
    print("Mean   :", round(mean, 2))
    print("Median :", round(median, 2))
    
    if mean > median:
        print("Interpretation: Distribution is likely right-skewed.")
    elif mean < median:
        print("Interpretation: Distribution is likely left-skewed.")
    else:
        print("Interpretation: Mean and median are approximately equal.")
```

```python
def iqr_analysis(data, column):
    
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    result = {
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier %': (len(outliers) / len(data)) * 100
    }
    
    return result, outliers
```

```python
sales_iqr, sales_outliers = iqr_analysis(df, 'Sales')

display(
    pd.DataFrame([sales_iqr]).round(2)
)
```

```python
profit_iqr, profit_outliers = iqr_analysis(df, 'Profit')

display(
    pd.DataFrame([profit_iqr]).round(2)
)
```

```python
outlier_summary = pd.DataFrame([
    sales_iqr,
    profit_iqr
])

display(outlier_summary.round(2))
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()

average_sales = df['Sales'].mean()
average_profit = df['Profit'].mean()
average_discount = df['Discount'].mean()

profit_margin = (
    total_profit / total_sales
) * 100

kpi = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Profit',
        'Total Quantity',
        'Average Sales',
        'Average Profit',
        'Average Discount',
        'Profit Margin'
    ],
    
    'Value': [
        total_sales,
        total_profit,
        total_quantity,
        average_sales,
        average_profit,
        average_discount,
        profit_margin
    ]
})

display(kpi.round(2))
```

```python
product_profit = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values()
)

print("10 Least Profitable Products")
display(
    product_profit.head(10).to_frame('Total Profit')
)

print("\n10 Most Profitable Products")
display(
    product_profit.tail(10)
    .sort_values(ascending=False)
    .to_frame('Total Profit')
)
```

```python
product_analysis = (
    df.groupby('Product Name')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Quantity=('Quantity', 'sum')
    )
)

high_sales_low_profit = product_analysis[
    (product_analysis['Total_Sales'] >
     product_analysis['Total_Sales'].quantile(0.75))
    &
    (product_analysis['Total_Profit'] < 0)
].sort_values(
    'Total_Sales',
    ascending=False
)

display(high_sales_low_profit.head(10))
```

```python
discount_analysis = (
    df.groupby('Discount')
    .agg(
        Average_Sales=('Sales', 'mean'),
        Average_Profit=('Profit', 'mean'),
        Total_Profit=('Profit', 'sum'),
        Number_of_Orders=('Profit', 'count')
    )
    .sort_index()
)

display(
    discount_analysis.round(2)
)
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=discount_analysis,
    x=discount_analysis.index,
    y='Average_Profit',
    marker='o'
)

plt.title('Average Profit by Discount Level')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.legend(['Chandan K Rao'])
plt.show()
```

```python
# ============================================================
# BUSINESS INSIGHTS & RECOMMENDATIONS
# ============================================================

from IPython.display import display, Markdown

display(Markdown("""

#  Business Insights & Recommendations

## 1️. Technology is the strongest-performing category

**Insight:**  
Technology is the strongest-performing product category, generating approximately **$836K in sales** and **$145K in profit**. Furniture generates around **$742K in sales**, but its profit is only about **$18K**.

**Business Recommendation:**  
The company should focus more on high-performing Technology products through better inventory planning, targeted marketing campaigns, and product expansion. At the same time, the pricing and cost structure of Furniture products should be reviewed.

---

## 2️. West Region has the highest sales performance

**Insight:**  
The **West region** generates the highest sales at approximately **$725K**, followed by East at **$679K**, Central at **$501K**, and South at **$392K**.

**Business Recommendation:**  
The company should study the successful products, customers, and sales strategies in the West region and apply similar strategies to improve sales performance in the South region.

---

## 3️. High sales do not always mean high profit

**Insight:**  
Product-level analysis identifies products that generate high sales but have **negative profit**. This indicates that revenue alone cannot be used to evaluate product performance.

**Business Recommendation:**  
Management should review loss-making products by analyzing:

- Product pricing
- Discount levels
- Supplier costs
- Shipping costs
- Profit margins

Products that continuously generate losses should be repriced, optimized, or discontinued.

---

## 4️. Sales are concentrated in high-performing sub-categories

**Insight:**  
The **Phones** sub-category is one of the strongest performers, generating approximately **$0.33M in sales**, followed by sub-categories such as Chairs and Storage.

**Business Recommendation:**  
The company should maintain sufficient inventory for high-performing sub-categories and use cross-selling and bundle offers to increase sales of related products.

---

## 5️. Profitability should be monitored along with sales

**Insight:**  
The analysis shows that high sales do not necessarily result in high profitability. Therefore, evaluating performance using sales alone can lead to incorrect business decisions.

**Business Recommendation:**  
Management should continuously monitor the following KPIs together:

- Total Sales
- Total Profit
- Profit Margin
- Total Orders
- Total Quantity
- Discount

This will provide a more complete picture of business performance.

---

#  Key Business Recommendations

### 1. Increase Technology Category Investment
Focus marketing, inventory, and product expansion on high-performing Technology products.

### 2. Improve South Region Performance
Analyze successful strategies from the West region and implement targeted campaigns in the South.

### 3. Control Loss-Making Products
Identify products with high sales but negative profit and optimize their pricing, discounts, and costs.

### 4. Optimize Product Inventory
Maintain adequate stock for high-performing sub-categories such as Phones, Chairs, and Storage.

### 5. Focus on Profitability
Use **Sales + Profit + Profit Margin** together instead of using sales alone to evaluate business performance.

---

#  Executive Summary

The Superstore analysis shows that **Technology is the strongest-performing category**, while the **West region leads regional sales**. However, some products generate high sales but negative profit, indicating the importance of monitoring profitability rather than revenue alone.

The business should focus on **Technology expansion, regional growth, product profitability, inventory optimization, discount management, and continuous KPI monitoring** to improve overall business performance.

"""))
```
